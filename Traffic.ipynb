```python
import pandas as pd
from sqlalchemy import create_engine,text
```

```python
import streamlit as st
```

```python
df=pd.read_csv(r"D:\DS\MiniProjects\Traffic_CrashesData.csv")
```

```python
#Connecting to the default 'mysql' database container 
base_engine = create_engine('mysql://root:password@localhost:3306/mysql')

#Creating the database if it doesn't exist
with base_engine.connect() as conn:
    conn.execute(text("CREATE DATABASE IF NOT EXISTS traffic_analysis"))
    conn.commit()  # Make sure the change is saved

#Connecting to the newly created database
engine = create_engine('mysql://root:password@localhost:3306/traffic_analysis')

#Pushing the dataframe
df.to_sql('traffic_analysis', con=engine, if_exists='replace', index=False)
```

```python
# 1. Force Pandas to stop breaking the table into pieces
#pd.set_option('display.expand_frame_repr', False)

# 2. Tell Pandas to show all columns without truncating them
#pd.set_option('display.max_columns', None)

from tabulate import tabulate
query1 = """SELECT 
    WEATHER_CONDITION,
    FIRST_CRASH_TYPE,
    COUNT(*) AS total_crashes
FROM 
    traffic_analysis
GROUP BY 
    WEATHER_CONDITION, 
    FIRST_CRASH_TYPE
ORDER BY 
    total_crashes DESC
LIMIT 5"""
df1 = pd.read_sql_query(query1, con=engine)
st.dataframe(df1, use_container_width=True)
```
